```python
import re
```

```python
with open('ecoli.csv', 'r') as f:
    lines = f.readlines()

with open('ecoli2.csv', 'w') as f:
    for line in lines:
        new_line = re.sub(r'\s+', ',', line.strip())
        f.write(new_line + '\n')
```
