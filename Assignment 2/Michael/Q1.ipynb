```python
# -------------------------------------------------------
# Lymphography Decision Tree Classification
# -------------------------------------------------------

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import OneHotEncoder
import matplotlib.pyplot as plt

# Load the lymphography dataset
# If you have a CSV file:
data = pd.read_csv("/content/lymph (3).csv")

# If loading from UCI website manually, make sure the CSV has 19 columns.
# I will assume your dataset is loaded into 'data'.

# -------- PREPROCESSING --------

# Last column = class attribute
X = data.iloc[:, :-1]   # first 18 attributes
y = data.iloc[:, -1]    # class labels

# One-hot encode all categorical features
encoder = OneHotEncoder(sparse_output=False)
X_encoded = encoder.fit_transform(X)

# Train-test split (60% train, 40% test)
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.40, random_state=42
)

# -------- DECISION TREE MODEL (min_samples_split = 4) --------

clf_4 = DecisionTreeClassifier(
    criterion="entropy",
    min_samples_split=4,
    random_state=42
)

clf_4.fit(X_train, y_train)

# Predict
y_pred_4 = clf_4.predict(X_test)

# Accuracy
accuracy_4 = accuracy_score(y_test, y_pred_4)
print("Accuracy (min_samples_split = 4):", accuracy_4)

# -------- PLOT TREE --------

plt.figure(figsize=(24, 12))
plot_tree(clf_4,
          filled=True,
          fontsize=6)
plt.title("Decision Tree (min_samples_split = 4)")
plt.show()
```

```python
# -------- DECISION TREE MODEL (min_samples_split = 32) --------

clf_32 = DecisionTreeClassifier(
    criterion="entropy",
    min_samples_split=32,
    random_state=42
)

clf_32.fit(X_train, y_train)

# Predict
y_pred_32 = clf_32.predict(X_test)

# Accuracy
accuracy_32 = accuracy_score(y_test, y_pred_32)
print("Accuracy (min_samples_split = 32):", accuracy_32)

# -------- PLOT TREE --------

plt.figure(figsize=(24, 12))
plot_tree(clf_32,
          filled=True,
          fontsize=6)
plt.title("Decision Tree (min_samples_split = 32)")
plt.show()
```
